### (~:>) fragment

```spiral
instance (~:>) fragment = fun x =>
    real
        typecase t with
        | array_base (view' (html_element ~el)) =>
            inl x = a x
            inl x = am.toList `a `int `(view' (html_element el)) x
            inl x = listm.map `(view' (html_element el)) `view (element_to_view `el) x
            view_list_to_fragment x
        | list (view' (html_element ~el)) =>
            inl x = listm.map `(view' (html_element el)) `view (element_to_view `el) x
            view_list_to_fragment x
        | list view =>
            view_list_to_fragment x
        | _ => x
```

### (~:>) view

```spiral
instance (~:>) view = fun x =>
    real
        typecase t with
        | view' (html_element _) => element_to_view x
        | _ => x
```

### typecheck_signal

```spiral
inl typecheck_signal forall (t : * -> *) u. (signal : t u) : () =
    real
        typecase t with
        | signal => ()
        | arc_signal => ()
        | rw_signal => ()
        | arc_rw_signal => ()
        | read_signal => ()
        | arc_read_signal => ()
        | write_signal => ()
        | arc_write_signal => ()
        | memo => ()
        | arc_memo => ()
        | _ => error_type `(()) ("invalid signal", ``(t u))
```

```spiral
inl tag_text (tag : string) props (children : string) : string =
    inl props = props |> concat_props
    $'"<" + !tag + " " + !props + ">" + !children + "</" + !tag + ">"'
```

### tag_raw

```spiral
inl tag_raw tag props children : string =
    inl children = children () |> fragment_to_view
    tag_text tag props $'"{ !children }"'
```

### reactive_view

```spiral
inl reactive_view (fn : () -> view) : view =
    inl fn = fn |> send_wrapper_new
    $'leptos::prelude::IntoAny::into_any(move || (*!fn)())'
```

### reactive_fragment

```spiral
inl reactive_fragment (fn : () -> fragment) : fragment =
    reactive_view (fun () => fn () |> fragment_to_view)
    |> view_to_fragment
```

### tag_element

```spiral
inl tag_element tag props children : view' (html_element _) =
    tag_raw tag props children
    |> macro_to_element
```

### tag_closed_raw

```spiral
inl tag_closed_raw tag props =
    inl tag : string = tag
    inl props = props |> concat_props
    $'"<" + !tag + " " + !props + " />"'
```

### tag_closed

```spiral
inl tag_closed tag props : view' (html_element _) =
    tag_closed_raw tag props
    |> macro_to_element
```

### for

### for

```spiral
inl for forall t u (signal : * -> *).
    (signal : signal (am'.vec t))
    (key_fn : t -> u)
    (children' : t -> fragment)
    : view
    =
    signal |> typecheck_signal
    inl items = (fun () => signal_get' signal : am'.vec t) |> send_wrapper_new
    inl key = (fun (x : send_wrapper t) => x |> send_wrapper_get |> key_fn) |> send_wrapper_new
    inl child = (fun (x : send_wrapper t) => x |> send_wrapper_get |> children' |> fragment_to_view) |> send_wrapper_new
    for [
        $'"each={ move || (*!items)().into_iter().map(leptos::__reexports::send_wrapper::SendWrapper::new).collect::<Vec<_>>() }"'
        $'"key={ move |x: &leptos::__reexports::send_wrapper::SendWrapper<_>| (*!key)(x.clone()) }"'
        $'"children={ move |x: leptos::__reexports::send_wrapper::SendWrapper<_>| (*!child)(x) }"'
    ]
```

### use_location

### use_navigate

```spiral
inl use_navigate () : string -> () =
    inl navigate : threading.arc (rust.dyn' (rust.action_fn2 (rust.ref sm'.str) navigate_options)) =
        !\($'"std::sync::Arc::new(leptos_router::hooks::use_navigate())"')
    fun url =>
        inl url = url |> sm'.as_str
        !\\((navigate, url), $'"$0($1, Default::default())"')
```

```spiral
inl url_pathname (url : url) : sm'.std_string =
    !\\(url, $'"$0.path().to_string()"')
```

### use_url

```spiral
inl use_url () =
    inl location = use_location ()

    fun () =>
        inl url_pathname = location |> location_pathname |> signal_get |> sm'.from_std_string
        inl url_search = location |> location_search |> signal_get |> sm'.from_std_string
        inl url_search =
            if url_search = ""
            then ""
            else "?" ++# url_search
        url_pathname ++# url_search
    |> new_arc_memo
```
