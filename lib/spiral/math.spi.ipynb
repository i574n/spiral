## fsharp

```spiral
:= atan2 (y : f64) (x : f64) : f64 =
    backend_switch {
        Gleam = \ () =>
            global """
                @external(erlang, "math", "atan2")
                fn spiral_math_atan2(y: Float, x: Float) -> Float
                """
            $'spiral_math_atan2(!y, !x)' : float
        Lua = \ () => $'math.atan2(!y, !x)' : float
        Fsharp = \ () => $'System.Math.Atan2 (!y, !x)' : float
        Python = \ () =>
            global "import math"
            $'math.atan2(!y, !x)' : float
        Rust = \ () => $'(!y).atan2(!x)' : float
        Zig = \ () => $'std.math.atan2(!y, !x)' : float
        Lean = \ () => $'Float.atan2 !y !x ' : float
        C = \ () =>
            global "#include <tgmath.h>"
            $'atan2(!y, !x)' : float
        Cpp = \ () =>
            global "#include <cmath>"
            $'std::atan2(!y, !x)' : float
        Pascal = \ () => $'ArcTan2(!y, !x)' : float
        Wasm = \ () => $'(call $math_atan2 !y !x)' : float
        TypeScript = \ () => $'Math.atan2(!y, !x)' : float
        Kotlin = \ () => $'kotlin.math.atan2(!y, !x)' : float
        Swift = \ () => $'atan2(!y, !x)' : float
    }
```

## floor

```spiral
:= floor \* t {float}. (n : t) : t =
    backend_switch {
        Gleam = \ () =>
            global "import gleam/float"
            n |> $'float.floor' : t
        Lua = \ () => n |> $'math.floor' : t
        Fsharp = \ () => n |> $'floor' : t
        Python = \ () =>
            global "import math"
            $'math.floor(!n)' : t
        Rust = \ () => $'!n.floor()' : t
        Zig = \ () => $'@@floor(!n)' : t
        Lean = \ () => $'(!n).floor' : t
        C = \ () =>
            global "#include <tgmath.h>"
            $'floor(!n)' : t
        Cpp = \ () =>
            global "#include <cmath>"
            $'std::floor(!n)' : t
        Pascal = \ () => $'Int(!n) - Ord(Frac(!n) < 0)' : t
        Wasm = \ () => $'(`t.floor !n)' : t
        TypeScript = \ () => $'Math.floor(!n)' : t
        Kotlin = \ () => $'kotlin.math.floor(!n)' : t
        Swift = \ () => $'(!n).rounded(.down)' : t
    }
```

```spiral
///- --test --symbol
///> rust
///> gleam
///> lua
///> cuda

0.6 |> floor
|> _assert_eq 0f64
```

## round

```spiral
:= round \* t {number; float}. (x : t) : t =
    := round_half_to_even () =
        := lower = floor x
        := difference = x - lower
        ? difference > 0.5 ?> lower + 1
        |? difference < 0.5 ?> lower
        |? floor (lower / 2) * 2 = lower ?> lower
        |: lower + 1
    backend_switch {
        Gleam = round_half_to_even
        Lua = round_half_to_even
        Fsharp = \ () => x |> $'round' : t
        Python = \ () => x |> $'round' : t
        Rust = \ () => $'(!x).round_ties_even()' : t
        Zig = round_half_to_even
        Lean = round_half_to_even
        C = round_half_to_even
        Cpp = round_half_to_even
        Pascal = round_half_to_even
        TypeScript = round_half_to_even
        Wasm = \ () => $'(`t.nearest !x)' : t
        Kotlin = \ () => $'kotlin.math.round(!x)' : t
        Swift = \ () => $'(!x).rounded(.toNearestOrEven)' : t
    }
```

```spiral
///- --test --symbol
///> rust
///> gleam
///> lua
///> cuda

0.5 |> round
|> _assert_eq 0f64

0.6 |> round
|> _assert_eq 1f64

1.5 |> round
|> _assert_eq 2f64

2.5 |> round
|> _assert_eq 2f64

3.5 |> round
|> _assert_eq 4f64
```

## log_base

```spiral
:= log_base (new_base : f64) (a : f64) : f64 =
    backend_switch {
        Gleam = \ () => log a / log new_base
        Lua = \ () => log a / log new_base
        Fsharp = \ () => $'System.Math.Log (!a, !new_base)' : f64
        Python = \ () => log a / log new_base
        Rust = \ () => $'(!a).log(!new_base)' : f64
        TypeScript = \ () => log a / log new_base
        Zig = \ () => log a / log new_base
        Lean = \ () => log a / log new_base
        Kotlin = \ () => log a / log new_base
        Swift = \ () => log a / log new_base
    }
```

```spiral
///- --test --symbol

100 |> log_base 10
|> _assert_eq 2
```
