## square

```spiral
inl square x =
    x ** 2
```

```spiral
///- --test

5f64
|> sqrt
|> square
|> _assert_approx_eq None 5
```

```spiral
///- --test

e () |> square
|> _assert_approx_eq None 7.3890560989306495
```

```spiral
///- --test

2 * 2 / 0.4f64 |> sqrt
|> _assert_approx_eq None 3.1622776601683795
```

```spiral
///- --test

2f64 / 3
|> _assert_approx_eq None 0.6666666666666666
```

```spiral
///- --test

2f64 |> log
|> _assert_approx_eq None 0.6931471805599453
```

```spiral
///- --test

pi
|> _assert_approx_eq None 3.141592653589793f64
```

## fsharp

### atan2

```spiral
inl atan2 (y : f64) (x : f64) : f64 =
    backend_switch {
        Gleam = fun () =>
            global """
                @external(erlang, "math", "atan2")
                fn spiral_math_atan2(y: Float, x: Float) -> Float
                """
            $'spiral_math_atan2(!y, !x)' : float
        Lua = fun () => $'math.atan2(!y, !x)' : float
        Fsharp = fun () => $'System.Math.Atan2 (!y, !x)' : float
        Python = fun () => 
            global "import math"
            $'math.atan2(!y, !x)' : float
        Rust = fun () => $'(!y).atan2(!x)' : float
        Zig = fun () => $'std.math.atan2(!y, !x)' : float
        Lean = fun () => $'Float.atan2 !y !x ' : float
        C = fun () =>
            global "#include <tgmath.h>"
            $'atan2(!y, !x)' : float
        Cpp = fun () =>
            global "#include <cmath>"
            $'std::atan2(!y, !x)' : float
        Delphi = fun () => $'ArcTan2(!y, !x)' : float
        Wasm = fun () => $'(call $math_atan2 !y !x)' : float
        TypeScript = fun () => $'Math.atan2(!y, !x)' : float
        Kotlin = fun () => $'kotlin.math.atan2(!y, !x)' : float
        Swift = fun () => $'atan2(!y, !x)' : float
    }
```

```spiral
///- --test
///> gleam
///> lua
///> fsharp
///> cuda

0 |> atan2 1
|> _assert_eq 1.5707963267948966
```

## floor

```spiral
inl floor forall t {float}. (n : t) : t =
    backend_switch {
        Gleam = fun () =>
            global "import gleam/float"
            n |> $'float.floor' : t
        Lua = fun () => n |> $'math.floor' : t
        Fsharp = fun () => n |> $'floor' : t
        Python = fun () => 
            global "import math"
            $'math.floor(!n)' : t
        Rust = fun () => $'!n.floor()' : t
        Zig = fun () => $'\@floor(!n)' : t
        Lean = fun () => $'(!n).floor' : t
        C = fun () =>
            global "#include <tgmath.h>"
            $'floor(!n)' : t
        Cpp = fun () =>
            global "#include <cmath>"
            $'std::floor(!n)' : t
        Delphi = fun () => $'Int(!n) - Ord(Frac(!n) < 0)' : t
        Wasm = fun () => $'(`t.floor !n)' : t
        TypeScript = fun () => $'Math.floor(!n)' : t
        Kotlin = fun () => $'kotlin.math.floor(!n)' : t
        Swift = fun () => $'(!n).rounded(.down)' : t
    }
```

```spiral
///- --test
///> gleam
///> lua
///> fsharp
///> cuda

0.6 |> floor
|> _assert_eq 0f64
```

## ceil

```spiral
inl ceil forall t {float}. (n : t) : t =
    backend_switch {
        Gleam = fun () =>
            global "import gleam/float"
            n |> $'float.ceiling' : t
        Lua = fun () => n |> $'math.ceil' : t
        Fsharp = fun () => n |> $'ceil' : t
        Python = fun () =>
            global "import math"
            n |> $'math.ceil' : t
        Rust = fun () => $'!n.ceil()' : t
        Zig = fun () => $'\@ceil(!n)' : t
        Lean = fun () => $'(!n).ceil' : t
        C = fun () =>
            global "#include <tgmath.h>"
            $'ceil(!n)' : t
        Cpp = fun () =>
            global "#include <cmath>"
            $'std::ceil(!n)' : t
        Delphi = fun () => $'Int(!n) + Ord(Frac(!n) > 0)' : t
        Wasm = fun () => $'(`t.ceil !n)' : t
        TypeScript = fun () => $'Math.ceil(!n)' : t
        Kotlin = fun () => $'kotlin.math.ceil(!n)' : t
        Swift = fun () => $'(!n).rounded(.up)' : t
    }
```

```spiral
///- --test
///> gleam
///> lua
///> fsharp
///> cuda

0.6 |> ceil
|> _assert_eq 1f64
```

## round

```spiral
inl round forall t {number; float}. (x : t) : t =
    inl round_half_to_even () =
        inl lower = floor x
        inl difference = x - lower
        if difference > 0.5 then lower + 1
        elif difference < 0.5 then lower
        elif floor (lower / 2) * 2 = lower then lower
        else lower + 1
    backend_switch {
        Gleam = round_half_to_even
        Lua = round_half_to_even
        Fsharp = fun () => x |> $'round' : t
        Python = fun () => x |> $'round' : t
        Rust = fun () => $'(!x).round_ties_even()' : t
        Zig = round_half_to_even
        Lean = round_half_to_even
        C = round_half_to_even
        Cpp = round_half_to_even
        Delphi = round_half_to_even
        TypeScript = round_half_to_even
        Wasm = fun () => $'(`t.nearest !x)' : t
        Kotlin = fun () => $'kotlin.math.round(!x)' : t
        Swift = fun () => $'(!x).rounded(.toNearestOrEven)' : t
    }
```

```spiral
///- --test
///> gleam
///> lua
///> fsharp
///> cuda

0.5 |> round
|> _assert_eq 0f64

0.6 |> round
|> _assert_eq 1f64

1.5 |> round
|> _assert_eq 2f64

2.5 |> round
|> _assert_eq 2f64

3.5 |> round
|> _assert_eq 4f64
```

## log_base

```spiral
inl log_base (new_base : f64) (a : f64) : f64 =
    backend_switch {
        Gleam = fun () => log a / log new_base
        Lua = fun () => log a / log new_base
        Fsharp = fun () => $'System.Math.Log (!a, !new_base)' : f64
        Python = fun () => log a / log new_base
        Rust = fun () => $'(!a).log(!new_base)' : f64
        TypeScript = fun () => log a / log new_base
        Zig = fun () => log a / log new_base
        Lean = fun () => log a / log new_base
        Kotlin = fun () => log a / log new_base
        Swift = fun () => log a / log new_base
    }
```

```spiral
///- --test

100 |> log_base 10
|> _assert_eq 2
```
