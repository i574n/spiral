# near_wallet

## near_wallet

```spiral
open lustre
```

### model

```spiral
type model = int
```

### init

```spiral
inl init initial_count : model =
    if initial_count < 0
    then 0
    else initial_count
```

### msg

```spiral
union msg =
    | Incr
    | Decr
```

### update

```spiral
inl update n (model, msg) =
    match msg with
    | Incr => model + n
    | Decr => model - n
```

### view

```spiral
inl view model =
    div [
        style [
            "display", "flex"
        ]
    ] [
        div [
            style [
                "display", "flex"
                "flex-direction", "column"
            ]
        ] [
            button [
                on_click Incr
            ] [
                text "+"
            ]
            p [
                style [
                    "text-align", "center"
                ]
            ] [
                model |> sm'.obj_to_string |> text
            ]
            button [
                on_click Decr
            ] [
                text "-"
            ]
        ]
    ]
```

## tests

```spiral
///- --test
///> gleam -t javascript -d lustre=\">=4.6.0\" lustre_ui=\"1.0.0-rc.1\" gleam_json=\">=2.0.0\"

inl model = init (dyn -3i32)
inl model = update 1i32 (model, Incr)
inl model = update 1i32 (model, Incr)
inl model = update 1i32 (model, Decr)
model |> view |> to_html |> expect_html "<div style=\"display:flex;\"><div style=\"display:flex;flex-direction:column;\"><button role=\"button\" tabindex=\"0\" class=\"lustre-ui-button\">+</button><p style=\"text-align:center;\">1</p><button role=\"button\" tabindex=\"0\" class=\"lustre-ui-button\">-</button></div></div>"
```

## main

```spiral
///> _

inl main' (n : int) =
    simple init (update 1) view
    |> start "#app_"

inl main () =
    $'!main' (0)' : ()
```
