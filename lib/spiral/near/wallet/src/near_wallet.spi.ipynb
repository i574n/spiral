```spiral
:< lustre

:== model = int

:= init initial_count : model =
    ? initial_count < 0
    ?> 0
    |: initial_count

:| msg =
    | Increment
    | Decrement

:= update n (model, msg) =
    ?= msg ?|
    | Increment => model + n
    | Decrement => model - n

:= view model =
    div [
        style [
            "display", "flex"
        ]
    ] [
        div [
            style [
                "display", "flex"
                "flex-direction", "column"
            ]
        ] [
            button [
                on_click Increment
            ] [
                text "+"
            ]
            p [
                style [
                    "text-align", "center"
                ]
            ] [
                model |> sm'::obj_to_string |> text
            ]
            button [
                on_click Decrement
            ] [
                text "-"
            ]
        ]
    ]
```

#### tests (test)

```spiral
///- --test --symbol
///> gleam -t javascript -d lustre=\">=5.7.0\" gleam_json=\">=2.0.0\"

:= model = init (dyn -3i32)
:= model = update 1i32 (model, Increment)
:= model = update 1i32 (model, Increment)
:= model = update 1i32 (model, Decrement)
model
|> view
|> to_html
|> expect_html (
    "<div style=\"display:flex;\"><div style=\"display:flex;flex-direction:column;\"><button "
    ^^ "class=\"lustre-ui-button\" role=\"button\" tabindex=\"0\">+</button><p "
    ^^ "style=\"text-align:center;\">1</p><button class=\"lustre-ui-button\" role=\"button\" "
    ^^ "tabindex=\"0\">-</button></div></div>"
)
```
