```spiral
:= concat_rev (acc : list string) : string =
    :~ ^ pass (rev : bool) (xs : list string) (out : list string) : list string =
        ?= xs ?|
        | a :: rest =>
            ?= rest ?|
            | b :: rest' => pass rev rest' ((? rev ?> b ++# a |: a ++# b) :: out)
            | [] => a :: out
        | [] => out
    :~ ^ 루프 (rev : bool) (xs : list string) : string =
        ?= xs ?|
        | [] => ""
        | x :: rest =>
            ?= rest ?|
            | [] => x
            | _ => 루프 (not rev) (pass rev xs [])
    ? is_rust () ?>
        ?= acc ?|
        | _ :: _ :: _ :: _ =>
            := total : i32 = (0i32, acc) ||> listm::fold \ n (x : string) => n + ##x
            := size : convert::unativeint = convert::convert total
            := buf : $'Rc<RefCell<Vec<u8>>>' = $'Rc::new(RefCell::new(vec\![0u8; !size]))'
            := _ : i32 = (total, acc) ||> listm::fold \ end (x : string) =>
                := start = end - ##x
                := from : convert::unativeint = convert::convert start
                := upto : convert::unativeint = convert::convert end
                $'!buf.borrow_mut()[!from..!upto].copy_from_slice(!x.as_bytes())' : ()
                start
            $'unsafe { Rc::<str>::from(std::str::from_utf8_unchecked(&!buf.borrow())) }' : string
        | _ => 루프 !+ acc
    |: 루프 !+ acc
```

```spiral
:= many_till \* b c. (p : parser b) (end_p : parser c) : parser _ = \ t, i, s =>
    trace_entry "parsing.many_till" \ () => { i t_len = ##t }
    :~ ^ 루프 (acc : list b) (i' : int) s' : result (_ * int * parser_state) _ =
        ?= end_p (t, i', s') ?|
        | Ok (_, i'', s'') => Ok (acc |> listm::rev, i'', s'')
        | Error _ =>
            ?= p (t, i', s') ?|
            | Ok (x, i'', s'') =>
                ? i'' <> i'
                ?> s'' |> 루프 (x :: acc) i''
                |: fail (\ _ => "parsing.many_till / inner parser succeeded without consuming text") i' s'
            | Error e => e |> Error
    s |> 루프 [] i
```

```spiral
:= attempt \* t. (p : parser t) : parser t = \ t, i, s =>
    ?= p (t, i, s) ?|
    | Ok (r, i', s') => Ok (r, i', s')
    | Error e => e |> Error
```

```spiral
:= look_ahead (p : parser _) : parser _ = \ t, i, s =>
    ?= p (t, i, s) ?|
    | Ok (x, _, _) => Ok (x, i, s)
    | Error e => e |> Error
```

```spiral
:= pipe2 \* a b c. (p1 : parser a) (p2 : parser b) (f : ((() -> a) * (() -> b)) -> c) : parser c = \ t, i, s =>
    ?= p1 (t, i, s) ?|
    | Error e => e |> Error
    | Ok (ra, i', s') =>
        p2 (t, i', s')
        |> resultm::map \ rb, i'', s'' =>
            f ((\ () => ra), (\ () => rb)), i'', s''
```

```spiral
:= new_lines1 () : parser () = \ t, i, (parser_state s' &= s) =>
    := i', s'' = s |> advance_newlines t i
    ? i' <> i
    ?> Ok ((), i', s'')
    |:
        fail
            \ t, i, (parser_state s') =>
                := preview_end = i + get_max_context () |> min s'::text_length
                "parsing.new_lines1 / expected at least one new line / "
                ++# sm'::show { rest = t |> slice i preview_end }
            i s
```
