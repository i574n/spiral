## parsing

### position

### range

```spiral
type range =
    {
        from : int
        to : int
    }

type position =
    {
        line : int
        col : int
    }
```

### parser_state

```spiral
nominal parser_state =
    {
        line_start : int
        position : position
        text_length : int
    }
```

### new_parser_state

```spiral
inl new_parser_state (line_start : i32) (line : i32) (col : i32) (text_length : i32) =
    { line_start position = { line col }; text_length }
```

### new_parser_state\''

```spiral
inl new_parser_state' text_length =
    new_parser_state 0 1 1 text_length |> parser_state
```

### parser_error

```spiral
type parser_error =
    {
        format : string * int * parser_state -> string
        i : int
        s : parser_state
    }
```

### parser

```spiral
type parser t = string * int * parser_state -> result (t * int * parser_state) parser_error
```

### fail

```spiral
inl fail (format : string * int * parser_state -> string) (i : int) (s : parser_state) : result _ parser_error =
    Error { format i s }
```

### error_message

```spiral
inl error_message (t : string) (e : parser_error) : string =
    e.format (t, e.i, e.s)
```

### trace_entry

```spiral
inl trace_entry_enabled () =
    false

inl trace_entry (name : string) (locals : () -> _) : () =
    if trace_entry_enabled () then
        inl name = dyn name
        inl locals : () -> string = dyn fun () => locals () |> sm'.format
        join trace Verbose (fun () => name) locals
```

### slice

```spiral
inl slice (start : int) (end : int) (t : string) : string =
    inl t = dyn t
    if start >= end then ""
    else
        t
        |> sm'.range (am'.Start start) (am'.End fun _ => end)
```

### parse

```spiral
inl parse forall t.
    (p : parser t)
    (t : string)
    (s : parser_state)
    : result ((() -> t) * (() -> string) * parser_state) (() -> string)
    =
    p (t, 0, s)
    |> resultm.map fun result, i, (parser_state s' as s) =>
        (fun () => result),
        (fun () => t |> slice i s'.text_length),
        s
    |> resultm.map_error fun e =>
        fun () => e |> error_message t
```

### update_char

```spiral
inl update_char (parser_state s) (c : char) : parser_state =
    match c with
    | '\n' =>
        { s with
            line_start = s.line_start + s.position.col
            position = { line = s.position.line + 1; col = 1 }
        }
    | _ =>
        { s with
            position = { line = s.position.line; col = s.position.col + 1 }
        }
    |> parser_state
```

### update_span

```spiral
inl update_span (t : string) (i : int) (len : int) (parser_state s' as s) : parser_state =
    inl endi = i + len
    let rec 루프 (i' : int) (last_nl : int) (count : int) =
        if i' >= endi
        then last_nl, count
        else
            match t |> sm'.index i' with
            | '\n' => 루프 (i' + 1) i' (count + 1)
            | _ => 루프 (i' + 1) last_nl count
    inl last_nl, n = 루프 i -1 0
    if n = 0
    then { s' with position = { line = s'.position.line; col = s'.position.col + len } }
    else
        { s' with
            line_start = last_nl + 1
            position =
                {
                    line = s'.position.line + n
                    col = endi - last_nl
                }
            text_length = s'.text_length
        }
    |> parser_state
```

### update

```spiral
inl update t s =
    s
    |> update_span t 0 ##t
```

### advance_while

```spiral
inl advance_while (fn : char -> bool) (t : string) (i : int) (parser_state s' as s)
    : int * parser_state
    =
    if i >= s'.text_length
    then i, s
    else
        inl i' = i |> sm'.span_from fn t
        inl i' =
            if i' > s'.text_length
            then s'.text_length
            else i'
        inl consumed = i' - i
        if consumed = 0
        then i, s
        else i', s |> update_span t i consumed
```

### run_parser

```spiral
inl run_parser p t =
    match p (t, 0, ##t |> new_parser_state') with
    | Ok (result, i, (parser_state s' as s)) => Ok (result, t |> slice i s'.text_length, s)
    | Error e => (fun () => e |> error_message t) |> Error
```

### any_char

```spiral
inl any_char () : parser char = fun t, i, (parser_state s' as s) =>
    if i >= s'.text_length then
        fail (fun _, _, s => "parsing.any_char / unexpected end of t / " ++# sm'.show { s }) i s
    else
        inl c = t |> sm'.index i
        Ok (c, i + 1, c |> update_char s)
```

```spiral
///- --test
///> gleam
///> fsharp
///> cuda
///> typescript

"abc"
|> run_parser (any_char ())
|> (resultm.get >> sm'.show)
|> _assert_eq (sm'.show ('a', "bc", new_parser_state 0 1 2 3))
```

### p_char

```spiral
inl p_char (c : char) : parser char =
    fun t, i, (parser_state ({ line_start position = { line col } } as s') as s) =>
        if i >= s'.text_length then
            fail (fun _, _, (parser_state s') => "parsing.p_char / unexpected end of text / " ++# sm'.show { c s' }) i s
        else
            inl got = t |> sm'.index i
            if got = c
            then Ok (c, i + 1, got |> update_char s)
            else
                fail
                    fun t, i, (parser_state { line_start position = { line col } }) =>
                        inl line_end = i |> sm'.span_from ((<>) '\n') t
                        inl end = (i + get_max_context ()) |> min line_end
                        inl line_slice = t |> slice line_start end
                        inl new_line =
                            inl len = ##line_slice
                            if len > 0 && (line_slice |> sm'.index (len - 1)) = '\n' then "" else "\n"
                        inl pointer_line = (" " |> sm'.replicate (col - 1)) ++# "^"
                        "parsing.p_char / "
                        ++# sm'.show { expected = c; line col }
                        ++# "\n" ++# line_slice
                        ++# new_line
                        ++# pointer_line ++# "\n"
                    i s
```

### any_string

```spiral
inl any_string len : parser string = fun t, i, (parser_state s' as s) =>
    if s'.text_length - i >= len
    then Ok (t |> slice i (i + len), i + len, s |> update_span t i len)
    else
        fail (fun _, _, s => "parsing.any_string / unexpected end of text / " ++# sm'.show { s }) i s
```

### skip_any_string

```spiral
inl skip_any_string len : parser () = fun t, i, (parser_state s' as s) =>
    if s'.text_length - i >= len
    then Ok ((), i + len, s |> update_span t i len)
    else
        fail (fun _, _, s => "parsing.skip_any_string / unexpected end of text / " ++# sm'.show { s }) i s
```

### skip_many

```spiral
inl skip_many forall t. (p : parser t) : parser () = fun t, i, s =>
    trace_entry "parsing.skip_many" fun () => { t_len = ##t; i s }
    let rec 루프 (i' : int) (s' : parser_state) =
        match p (t, i', s') with
        | Error _ => Ok ((), i', s')
        | Ok (_, i'', s'') =>
            if i'' <> i'
            then s'' |> 루프 i''
            else fail (fun _ => "parsing.skip_many / inner parser consumed no text") i' s'
    s |> 루프 i
```

### skip_many1

```spiral
inl skip_many1 forall t. (a : parser t) : parser () = fun t, i, s =>
    trace_entry "parsing.skip_many1" fun () => { t_len = ##t; i s }
    match a (t, i, s) with
    | Error e => e |> Error
    | Ok (_, i', s') =>
        if i' <> i
        then skip_many a (t, i', s')
        else fail (fun _ => "parsing.skip_many1 / inner parser consumed no text") i' s'
```

### (>>.)

```spiral
inl (>>.) forall t u. (a : parser t) (b : parser u) : parser u = fun t, i, s =>
    match a (t, i, s) with
    | Ok (_, i', s') => b (t, i', s')
    | Error e => e |> Error
```

### (.>>)

```spiral
inl (.>>) forall t u. (a : parser t) (b : parser u) : parser t = fun t, i, s =>
    match a (t, i, s) with
    | Error e => e |> Error
    | Ok (ra, i', s') =>
        b (t, i', s')
        |> resultm.map fun _, i'', s'' =>
            ra, i'', s''
```

### (.>>.)

```spiral
inl (.>>.) forall t u. (a : parser t) (b : parser u) : parser ((() -> t) * (() -> u)) = fun t, i, s =>
    match a (t, i, s) with
    | Error e => e |> Error
    | Ok (ra, i', s') =>
        b (t, i', s')
        |> resultm.map fun rb, i'', s'' =>
            ((fun () => ra), (fun () => rb)), i'', s''
```

### (>>%)

### none_of

```spiral
inl none_of (chars : list char) : parser char = fun t, i, (parser_state s' as s) =>
    inl chars' () =
        chars |> listm'.box |> listm'.to_array' |> sm'.show
    if i >= s'.text_length then
        fail (fun _, _, s => "parsing.none_of / unexpected end of text / " ++# sm'.show { chars' = chars' (); s }) i s
    else
        inl ch = t |> sm'.index i
        if chars |> listm'.exists' ((=) ch) |> not
        then Ok (ch, i + 1, ch |> update_char s)
        else
            fail
                fun t, i, s =>
                    "parsing.none_of / unexpected char / "
                    ++# sm'.show { first_char = t |> sm'.index i; chars' = chars' (); s }
                i s
```

### many

```spiral
inl many forall b c.
    (p : _ * _ * _ -> _ (b * _ * _) c)
    : parser (list _)
    =
    fun t, i, s =>
        trace_entry "parsing.many" fun () => { t_len = ##t }
        let rec 루프 (acc : list b) (i' : int) (s' : parser_state)
            : result (list b * int * parser_state) _
            =
            match p (t, i', s') with
            | Error _ => Ok (acc |> listm.rev, i', s')
            | Ok (x, i'', s'') when i'' > i' => s'' |> 루프 (x :: acc) i''
            | Ok _ => fail (fun _ => "parsing.many / inner parser succeeded without consuming text") i' s'
        s |> 루프 [] i
```

### many1_satisfy

```spiral
inl many1_satisfy (fn : char -> bool) : parser string = fun t, i, (parser_state s' as s) =>
    let rec 루프 (i' : int) (s'' : parser_state) : int * parser_state =
        if i' >= s'.text_length
        then i', s''
        else
            inl c = t |> sm'.index i'
            if fn c
            then 루프 (i' + 1) (c |> update_char s'')
            else i', s''
    inl i', s'' = 루프 i s
    if i' > i
    then Ok (t |> slice i i', i', s'')
    else
        fail (fun _, i, s => "parsing.many1_satisfy / no matching char / " ++# sm'.show { i s }) i s
```

### advance_while_no_nl

```spiral
inl advance_while_no_nl (fn : char -> bool) (t : string) (i : int) (parser_state s' as s)
    : int * parser_state
    =
    if i >= s'.text_length
    then i, s
    else
        inl i' = i |> sm'.span_from fn t
        inl i' =
            if i' > s'.text_length
            then s'.text_length
            else i'
        inl consumed = i' - i
        if consumed = 0
        then i, s
        else
            i',
            { s' with position = { line = s'.position.line; col = s'.position.col + consumed } }
            |> parser_state
```

### advance_newlines

```spiral
inl advance_newlines (t : string) (i : int) (parser_state s' as s)
    : int * parser_state
    =
    if i >= s'.text_length
    then i, s
    else
        inl i' = i |> sm'.span_from ((=) '\n') t
        inl i' =
            if i' > s'.text_length
            then s'.text_length
            else i'
        inl consumed = i' - i
        if consumed = 0
        then i, s
        else
            i',
            { s' with
                line_start = i'
                position = { line = s'.position.line + consumed; col = 1 }
            }
            |> parser_state
```

### many_chars_till

```spiral
inl many_chars_till (p : parser char) (end_p : parser _) = fun t, i, (parser_state s' as s) =>
    trace_entry "parsing.many_chars_till" fun () => { t_len = ##t; i s }
    let rec 루프 (i' : int) (s'' : parser_state) : result (string * int * parser_state) parser_error =
        if i' >= s'.text_length
        then Ok ((if i' > i then t |> slice i i' else ""), i', s'')
        else
            inl i'', s''' = s'' |> advance_while_no_nl (fun c => c <> '\n') t i'
            if i'' >= s'.text_length
            then Ok ((if i'' > i then t |> slice i i'' else ""), i'', s''')
            else
                match end_p (t, i'', s''') with
                | Ok _ => Ok ((if i'' > i then t |> slice i i'' else ""), i'', s''')
                | Error _ =>
                    match p (t, i'', s''') with
                    | Error _ => Ok ((if i'' > i then t |> slice i i'' else ""), i'', s''')
                    | Ok (_, i''', s'''') when i''' > i'' => s'''' |> 루프 i'''
                    | Ok _ =>
                        fail (fun _ => "parsing.many_chars_till / inner parser succeeded without consuming text") i'' s'''
    s |> 루프 i
```

### concat_rev

```spiral
inl concat_rev (acc : list string) : string =
    let rec pass (rev : bool) (xs : list string) (out : list string) : list string =
        match xs with
        | a :: rest =>
            match rest with
            | b :: rest' => pass rev rest' ((if rev then b ++# a else a ++# b) :: out)
            | [] => a :: out
        | [] => out
    let rec 루프 (rev : bool) (xs : list string) : string =
        match xs with
        | [] => ""
        | x :: rest =>
            match rest with
            | [] => x
            | _ => 루프 (not rev) (pass rev xs [])
    if is_rust () then
        match acc with
        | _ :: _ :: _ :: _ =>
            inl total : i32 = (0i32, acc) ||> listm.fold fun n (x : string) => n + ##x
            inl buf : $'Rc<RefCell<Vec<u8>>>' = $'Rc::new(RefCell::new(vec\![0u8; !total as usize]))'
            inl _ : i32 = (total, acc) ||> listm.fold fun end (x : string) =>
                inl start = end - ##x
                $'!buf.borrow_mut()[!start as usize..!end as usize].copy_from_slice(!x.as_bytes())' : ()
                start
            $'unsafe { Rc::<str>::from(std::str::from_utf8_unchecked(&!buf.borrow())) }' : string
        | _ => 루프 true acc
    else 루프 true acc
```

```spiral
inl many1_strings (p : parser _) : parser string = fun t, i, s =>
    trace_entry "parsing.many1_strings" fun () => { t_len = ##t }
    match p (t, i, s) with
    | Error e => e |> Error
    | Ok (r, i', s') =>
        inl first = string_of r
        let rec 루프 (acc : list string) (i'' : int) s'' : result (string * int * _) _ =
            match p (t, i'', s'') with
            | Error _ =>
                inl r =
                    match acc with
                    | [] => first
                    | _ => first ++# (acc |> concat_rev)
                Ok (r, i'', s'')
            | Ok (r', i''', s''') when i''' > i'' => s''' |> 루프 (string_of r' :: acc) i'''
            | Ok _ => fail (fun _ => "parsing.many1_strings / inner parser succeeded without consuming text") i'' s''
        s' |> 루프 [] i'
```

### many_strings

```spiral
inl many_strings (p : parser _) : parser string = fun t, i, s =>
    trace_entry "parsing.many_strings" fun () => { t_len = ##t }
    match p (t, i, s) with
    | Error _ => Ok ("", i, s)
    | Ok (_, i', _) when i' = i => fail (fun _ => "parsing.many_strings / first inner parser consumed no text") i s
    | Ok (r, i', s') when is_rust () =>
        inl buf : $'Rc<RefCell<std::string::String>>' = $'Rc::new(RefCell::new(std::string::String::new()))'
        inl push (x : string) : () = $'!buf.borrow_mut().push_str(&*!x)'
        inl first = string_of r
        let rec 루프 (n : int) (i' : int) s' : result (string * int * _) _ =
            match p (t, i', s') with
            | Error _ =>
                inl r = if n = 0 then first else $'Rc::<str>::from(!buf.borrow().as_str())' : string
                Ok (r, i', s')
            | Ok (r', i'', s'') when i'' > i' =>
                if n = 0 then push first
                push (string_of r')
                s'' |> 루프 (n + 1) i''
            | Ok _ => fail (fun _ => "parsing.many_strings / inner parser succeeded without consuming text") i' s'
        s' |> 루프 0 i'
    | Ok (r, i', s') =>
        let rec 루프 (acc : list string) (i' : int) s' : result (string * int * _) _ =
            match p (t, i', s') with
            | Error _ =>
                inl r = string_of r
                inl r =
                    match acc with
                    | [] => r
                    | _ => r ++# (acc |> concat_rev)
                Ok (r, i', s')
            | Ok (r', i'', s'') when i'' > i' => s'' |> 루프 (string_of r' :: acc) i''
            | Ok _ => fail (fun _ => "parsing.many_strings / inner parser succeeded without consuming text") i' s'
        s' |> 루프 [] i'
```

### between

```spiral
inl between p_open p_close p_content : parser _ = fun t, i, s =>
    trace_entry "parsing.between" fun () => { t_len = ##t; i s }
    match p_open (t, i, s) with
    | Ok (_, i', s') =>
        match p_content (t, i', s') with
        | Ok (result, i'', s'') =>
            match p_close (t, i'', s'') with
            | Ok (_, i''', s''') => Ok (result, i''', s''')
            | Error e' =>
                fail
                    fun t, _, _ =>
                        inl e = fun () => e' |> error_message t
                        inl t_len = ##t
                        inl end' = (i' + get_max_context ()) |> min t_len
                        inl end'' = (i'' + get_max_context ()) |> min t_len
                        inl rest' = t |> slice i' end'
                        inl rest'' = t |> slice i'' end''
                        "parsing.between / expected closing delimiter / " ++# sm'.show { e t rest' rest'' }
                    i'' s''
        | Error _ =>
            match p_close (t, i', s') with
            | Ok (_, i''', s''') => Ok ("", i''', s''')
            | Error _ =>
                fail
                    fun t, i', _ =>
                        inl end' = (i' + get_max_context ()) |> min ##t
                        inl rest' = t |> slice i' end'
                        "parsing.between / expected content or closing delimiter / " ++# sm'.show { rest' }
                    i' s'
    | Error e => e |> Error
```

### sep_by

```spiral
inl sep_by forall b. p sep : parser (list b) = fun t, i, s =>
    trace_entry "parsing.sep_by" fun () => { t_len = ##t; i s }
    match p (t, i, s) with
    | Error _ => Ok ([], i, s)
    | Ok (first, i', s') =>
        let rec 루프 (acc : list b) (i'' : int) s'' : result (list b * int * parser_state) _ =
            match sep (t, i'', s'') with
            | Error _ => Ok (acc |> listm.rev, i'', s'')
            | Ok (_, i''', s''') =>
                if i''' = i'' then
                    fail (fun _ => "parsing.sep_by / separator consumed no text") i'' s''
                else
                    match p (t, i''', s''') with
                    | Ok (x, i'''', s'''') => s'''' |> 루프 (x :: acc) i''''
                    | Error _ => Ok (acc |> listm.rev, i'', s'')
        s' |> 루프 [ first ] i'
```

### sep_end_by

```spiral
inl sep_end_by forall b. p sep : parser (list b) = fun t, i, s =>
    trace_entry "parsing.sep_end_by" fun () => { t_len = ##t; i s }
    let rec 루프 acc i' s' =
        match p (t, i', s') with
        | Error _ => Ok (acc |> listm.rev, i', s')
        | Ok (x, i'', s'') =>
            if i'' = i' then
                fail (fun _ => "parsing.sep_end_by / element parser consumed no text") i' s'
            else
                inl acc' = x :: acc
                match sep (t, i'', s'') with
                | Error _ => Ok (acc' |> listm.rev, i'', s'')
                | Ok (_, i''', s''') =>
                    if i''' <> i''
                    then s''' |> 루프 acc' i'''
                    else fail (fun _ => "parsing.sep_end_by / separator consumed no text") i'' s''
    s |> 루프 [] i
```

### p_string

```spiral
inl p_string (str : string) : parser string = fun t, i, (parser_state s' as s) =>
    inl len = ##str
    if s'.text_length - i < len then
        fail (fun _, _, s => "parsing.p_string / unexpected end of text / " ++# sm'.show { expected = str; s }) i s
    else
        let rec eq k =
            if k >= len
            then true
            else t |> sm'.index (i + k) = str |> sm'.index k && eq (k + 1)
        if eq 0
        then Ok (t |> slice i (i + len), i + len, s |> update_span t i len)
        else
            fail
                fun t, i, (parser_state s' as s) =>
                    inl preview_end = (i + get_max_context ()) |> min s'.text_length
                    inl rest = t |> slice i preview_end
                    inl got = t |> slice i (i + len)
                    "parsing.p_string / unexpected string / "
                    ++# sm'.show { expected = str; got rest s }
                i s
```

### new_line

```spiral
inl new_line () : parser char = fun t, i, (parser_state s' as s) =>
    if i >= s'.text_length then
        fail (fun _, _, s => "parsing.new_line / unexpected end of text / " ++# sm'.show { s }) i s
    else
        inl c = t |> sm'.index i
        if c = '\n'
        then Ok (c, i + 1, c |> update_char s)
        else
            fail
                fun t, i, (parser_state s' as s) =>
                    inl preview_end = (i + get_max_context ()) |> min s'.text_length
                    inl rest = t |> slice i preview_end
                    "parsing.new_line / expected new line char / "
                    ++# sm'.show { first_char = t |> sm'.index i; rest s }
                i s
```

```spiral
///- --test

"hello\nworld"
|> run_parser (many_chars_till (any_char ()) (new_line ()))
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show ("hello", "\nworld", new_parser_state 0 1 6 11))
```

### many_till

```spiral
inl many_till forall b c. (p : parser b) (end_p : parser c) : parser _ = fun t, i, s =>
    trace_entry "parsing.many_till" fun () => { i t_len = ##t }
    let rec 루프 (acc : list b) (i' : int) s' : result (_ * int * parser_state) _ =
        match end_p (t, i', s') with
        | Ok (_, i'', s'') => Ok (acc |> listm.rev, i'', s'')
        | Error _ =>
            match p (t, i', s') with
            | Ok (x, i'', s'') =>
                if i'' <> i'
                then s'' |> 루프 (x :: acc) i''
                else fail (fun _ => "parsing.many_till / inner parser succeeded without consuming text") i' s'
            | Error e => e |> Error
    s |> 루프 [] i
```

```spiral
///- --test

"hello\nworld"
|> run_parser (many_till (any_char ()) (new_line ()))
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show (['h'; 'e'; 'l'; 'l'; 'o'], "world", new_parser_state 6 2 1 11))
```

### p_return

```spiral
inl p_return forall t. (x : t) : parser t = fun t, i, s =>
    Ok (x, i, s)
```

```spiral
///- --test

"abc"
|> run_parser (p_return 42i32)
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show (42i32, "abc", new_parser_state 0 1 1 3))
```

### attempt

```spiral
inl attempt forall t. (p : parser t) : parser t = fun t, i, s =>
    match p (t, i, s) with
    | Ok (r, i', s') => Ok (r, i', s')
    | Error e => e |> Error
```

```spiral
///- --test

"aabbcc"
|> run_parser (attempt (p_char 'a' >>. p_char 'a' >>. p_char 'b' >>. p_char 'c'))
|> resultm.map_error eval
|> _assert_eq (Error "parsing.p_char / { expected = c; line = 1; col = 4 }\naabbcc\n   ^\n")
```

### look_ahead

```spiral
inl look_ahead (p : parser _) : parser _ = fun t, i, s =>
    match p (t, i, s) with
    | Ok (x, _, _) => Ok (x, i, s)
    | Error e => e |> Error
```

```spiral
///- --test

"123"
|> run_parser (look_ahead (many1_chars (p_digit ())))
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show ("123", "123", new_parser_state 0 1 1 3))
```

### pipe2

```spiral
inl pipe2 forall a b c. (p1 : parser a) (p2 : parser b) (f : ((() -> a) * (() -> b)) -> c) : parser c = fun t, i, s =>
    match p1 (t, i, s) with
    | Error e => e |> Error
    | Ok (ra, i', s') =>
        p2 (t, i', s')
        |> resultm.map fun rb, i'', s'' =>
            f ((fun () => ra), (fun () => rb)), i'', s''
```

```spiral
///- --test

"pipe2"
|> run_parser (pipe2 (p_string "pipe") (spaces () >>. p_string "2") (fun a, b => a () ++# b ()))
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show ("pipe2", "", new_parser_state 0 1 6 5))
```

### new_lines

```spiral
inl new_lines () : parser () = fun t, i, s =>
    inl i', s' = s |> advance_newlines t i
    Ok ((), i', s')
```

```spiral
///- --test

"\n\nline1\nline2\n\nline4"
|> run_parser (many_till (any_char ()) (new_lines ()))
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show (([] : _ char), "line1\nline2\n\nline4", new_parser_state 2 3 1 20))
```

### new_lines1

```spiral
inl new_lines1 () : parser () = fun t, i, (parser_state s' as s) =>
    inl i', s'' = s |> advance_newlines t i
    if i' <> i
    then Ok ((), i', s'')
    else
        fail
            fun t, i, (parser_state s') =>
                inl preview_end = i + get_max_context () |> min s'.text_length
                "parsing.new_lines1 / expected at least one new line / "
                ++# sm'.show { rest = t |> slice i preview_end }
            i s
```

```spiral
///- --test

"\n\n\n"
|> run_parser (new_lines1 ())
|> resultm.get
|> sm'.show
|> _assert_eq (sm'.show ("", new_parser_state 3 4 1 3))
```
