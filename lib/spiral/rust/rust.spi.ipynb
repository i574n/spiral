### fix_closure

```spiral
inl fix_closure forall t. (depth : u8 * u8) (x : t) =
    backend_switch {
        Rust = fun () =>
            inl is_unit : bool =
                real
                    typecase t with
                    | () => true
                    | _ => false
            if is_unit
            then (!\($'"true; () }); //"') : bool) |> ignore
            else ($'true; (!x) }); //' : bool) |> ignore
    }
```

### loop

```spiral
inl loop (depth : i32) (fn : () -> ()) : () =
    (!\($'"true; loop { // rust.loop 1"') : bool) |> ignore
    fn ()

    listm.init depth id
    |> listm.iter fun n =>
        (!\($'"true; } // rust.loop 2 / n: {!n}"') : bool) |> ignore

    (!\($'"true; } // rust.loop 3"') : bool) |> ignore

    listm.init depth id
    |> listm.iter fun n =>
        (!\($'"true; { // rust.loop 4 / n: {!n}"') : bool) |> ignore
```
